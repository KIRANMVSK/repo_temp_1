```python
#now lest buld the real linear model

from torch.utils.data import TensorDataset,DataLoader
from torch.nn import MSELoss,Tanh,ReLU
from torch.optim import Adam
from torch import nn

#now lets crete x and y man
x=torch.tensor([[1,3,4],[2,4,5],[3,5,6],[4,5,6],[5,6,8],[1,3,4],[2,4,3],[3,4,5],[4,5,6],[5,5,6]])
y=torch.tensor([200,300,240,345,560,234,456,324,345,678])

#lets create them as the tensordataset
data=TensorDataset(x,y)

#creating dataloader
loader=DataLoader(data,batch_size=2,shuffle=True)

#now we nend create the instances for loss function , optimizer , activation
#it is linear so lets use the mse
loss=MSELoss()
relu=ReLU()
tanh=Tanh()
#optimizer=Adam(lr=0.02, params=)

   #now layers
cf1=nn.Linear(3,5)
cf2=nn.Linear(5,3)
cf3=nn.Linear(3,1)

optimizer=Adam(lr=0.02, params=[cf1.model_parameters(),cf2.model_parameters(),cf3.model_parameters()])

#now lets creae the moddel arrrangement

def forward(input):
  hidden=cf1(input)
  act1=relu(hidden)
  hidden=cf2(act1)
  act2=tanh(hidden)
  output=cf3(act2)
  return output

def train(epochs):
  for i in range(epochs):
    #batches in that epoch as loader gives one batch at a time
    for x_train,y_train in loader:
      x,y=x_train,y_train
      output=forward(x)

      #now predicting
      loss=loss(output,y)
      loss.backward()
      optimizer.step()
      optimizer.zero_grad()

train(3)




```

```python
'''1. Slicing in nn.Sequential

Suppose:

model = nn.Sequential(
    nn.Linear(4,16),  # 0
    nn.ReLU(),        # 1
    nn.Linear(16,8),  # 2
    nn.ReLU(),        # 3
    nn.Linear(8,3)    # 4
)

Access single layer:

model[0]
model[2]

You can also slice:

model[:2]

Output:

Sequential(
  (0): Linear(4,16)
  (1): ReLU()
)
model[:-1]

Output:

Sequential(
  (0): Linear(4,16)
  (1): ReLU()
  (2): Linear(16,8)
  (3): ReLU()
)

Very useful for removing classifier heads.

Example:

backbone = model[:-1]

Now:

x = backbone(x)

runs everything except the last layer.

This is a common transfer learning trick.

2. Freezing with slices

Freeze first 3 modules:

for layer in model[:3]:
    for p in layer.parameters():
        p.requires_grad = False

Freeze everything except last layer:

for layer in model[:-1]:
    for p in layer.parameters():
        p.requires_grad = False

Very clean.

3. Custom layer names inside Sequential

Instead of:

model[0]
model[2]
model[4]

you can give names.

Use OrderedDict:

from collections import OrderedDict

model = nn.Sequential(
    OrderedDict([
        ("fc1", nn.Linear(4,16)),
        ("relu1", nn.ReLU()),
        ("fc2", nn.Linear(16,8)),
        ("relu2", nn.ReLU()),
        ("classifier", nn.Linear(8,3))
    ])
)

Now:

model.fc1

❌ Doesn't work.

Sequential doesn't create attributes.

Instead:

model["fc1"]
model["classifier"]

✅ Works.

Check names:

for name, module in model.named_children():
    print(name)

Output:

fc1
relu1
fc2
relu2
classifier
4. Real-world approach: use nn.Module

Once models get bigger, people stop using numeric indexing.

Instead:

class MyNet(nn.Module):

    def __init__(self):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(4,16),
            nn.ReLU(),
            nn.Linear(16,8)
        )

        self.classifier = nn.Linear(8,3)

    def forward(self,x):

        x = self.encoder(x)
        x = self.classifier(x)

        return x

Now you get beautiful names:

model.encoder
model.classifier

Freeze encoder:

for p in model.encoder.parameters():
    p.requires_grad = False

Replace classifier:

model.classifier = nn.Linear(8,5)

This is exactly how transformer models are structured:

model.encoder
model.decoder
model.classifier
model.lm_head
model.embeddings'''

print('see how to write convential names')
```
